```python
import pandas as pd

df = pd.read_csv("ecommerce_uk_customers.csv")

df.head()
```

```python
df.info()
```

```python
# Select multiple columns
df[["customer_name", "age"]]
```

```python
# Basic Analysis
total_amount = df['total_spent'].sum()

print("Total Amount spent: ",total_amount)
```

```python
total_orders = df['orders_count'].sum()

print("Total Orders:", total_orders)
```

```python
average_amount = df['total_spent'].mean()

print("Average amount spent:", average_amount)
```

```python
# Sales greater than the average

high_sales = df[df['total_spent'] > 4900] #Filtring

high_sales
```

```python
# Maximum
# print(df['total_spent'].max())
df[df['total_spent'] == df['total_spent'].max()]
```

```python
# Minimum

df[df['total_spent'] == df['total_spent'].min()]
```

```python
# Sales per region

df['region'].value_counts()
```

```python
# number of sales per city

df['city'].value_counts()
```

```python
# Total spent per region

region_sales = df.groupby("region")[['total_spent', 'orders_count']].sum()

print(region_sales)
```

```python
# Total Orders per region

region_sales = df.groupby("city")[['total_spent', 'orders_count']].sum()

print(region_sales)
```

```python
# Sales per Category

category_sales = df.groupby('preferred_category')['total_spent'].sum()

category_sales
```

```python
# Sorting (ASC)

category_sales.sort_values()
```

```python
# Sorting (DESC)

category_sales.sort_values(ascending=False)
```

```python
# Top 5 Cities by total amount spent

city_sales = df.groupby("city")['total_spent'].sum()

city_sales.sort_values(ascending=False).head(3)
```

```python
# Bottom 3 Regions by Number of Orders

region_orders = df.groupby("region")['orders_count'].sum()

region_orders.sort_values().head(3)
```
